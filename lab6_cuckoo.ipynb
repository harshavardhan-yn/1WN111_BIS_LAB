```python

import random

# Step 1: Initialize subjects and time slots

subjects = [
    "Python",
    "Java",
    "DBMS",
    "Operating Systems"
]

time_slots = [
    "9-10 AM",
    "10-11 AM",
    "11-12 PM",
    "1-2 PM",
    "2-3 PM"
]

N = 10                  # Number of nests
Pa = 0.25               # Discovery probability
max_iterations = 100

# Step 2: Generate a random timetable

def generate_timetable():
    return [
        random.randint(0, len(time_slots) - 1)
        for _ in subjects
    ]

# Step 3: Calculate fitness (number of clashes)

def fitness(timetable):
    clashes = 0

    for i in range(len(timetable)):
        for j in range(i + 1, len(timetable)):

            if timetable[i] == timetable[j]:
                clashes += 1

    return clashes

# Step 4: Generate new solution using Levy Flight

def levy_flight(timetable):

    new_timetable = timetable.copy()

    # Heavy-tailed random step size
    step = int(random.paretovariate(1.5))

    step = max(1, min(step, len(subjects)))

    # Change randomly selected subject slots
    positions = random.sample(
        range(len(subjects)), step
    )

    for pos in positions:
        new_timetable[pos] = random.randint(
            0, len(time_slots) - 1
        )

    return new_timetable

# Step 5: Initialize random nests

nests = [generate_timetable() for _ in range(N)]

# Find the best initial solution
best = min(nests, key=fitness)

# Step 6: Main Cuckoo Search loop

for iteration in range(max_iterations):

    for i in range(N):

        # Generate a new cuckoo solution
        new_solution = levy_flight(nests[i])

        # Select a random host nest
        j = random.randint(0, N - 1)

        # Compare fitness
        if fitness(new_solution) < fitness(nests[j]):
            nests[j] = new_solution

    # Sort nests based on fitness
    nests.sort(key=fitness)

    # Step 7: Abandon worst nests
    abandon_count = int(Pa * N)

    for i in range(N - abandon_count, N):
        nests[i] = generate_timetable()

    # Update best solution
    current_best = min(nests, key=fitness)

    if fitness(current_best) < fitness(best):
        best = current_best.copy()

    # Stop if no clashes
    if fitness(best) == 0:
        break

# Step 8: Display final timetable

print("\nBEST TIMETABLE")
print("-------------------------")

for i in range(len(subjects)):
    slot = best[i]

    print(subjects[i], ":", time_slots[slot])

print("-------------------------")
print("Total clashes:", fitness(best))
print("Iterations:", iteration + 1)
```
